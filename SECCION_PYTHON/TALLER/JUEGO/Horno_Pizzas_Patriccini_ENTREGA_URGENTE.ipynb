```python
# Hemos puesto "import random" porque necesitamos la librería random para hacer cosas al azar.
# Queremos mezclar la lista de clientes y elegir qué pizza pide cada uno. Sin ella, el juego sería siempre igual.
import random
# Hemos puesto "import time" porque necesitamos la librería time para dos cosas:
# medir los 50 segundos que tiene el jugador y hacer la pausa de un segundo en la cuenta atrás del horno.
import time
# display y HTML permiten mostrar tarjetas con colores y negritas en Jupyter.
# clear_output limpia la salida anterior cuando volvemos a ejecutar la partida.
from IPython.display import display, HTML, clear_output  

# Usamos un diccionario: cada nombre de pizza (clave) está asociado a su lista de ingredientes (valor).
# Esto nos permite consultar fácilmente la receta correcta del pedido.
recetas = {
    "Margarita": ["masa", "tomate", "queso"],
    "Pepperoni": ["masa", "tomate", "queso", "pepperoni"],
    "4 quesos": ["masa", "tomate", "mix 4 quesos"],
    "Jamón y queso": ["masa", "tomate", "queso", "jamón"]
}
# Hemos creado este diccionario para relacionar cada número con un ingrediente.
# Así el jugador puede elegir los ingredientes escribiendo números en lugar de sus nombres.
ingredientes = {1: "masa", 2: "tomate", 3: "queso", 4: "mix 4 quesos", 5: "jamón", 6: "pepperoni"}  
# Hemos creado este diccionario para relacionar cada ingrediente con su emoji.
# Lo usamos para que los ingredientes se vean de forma más visual durante la partida.
iconos = {"masa": "🫓", "tomate": "🍅", "queso": "🧀", "mix 4 quesos": "🧀", "jamón": "🥓", "pepperoni": "🍕"}  


# Hemos creado la función "def tarjeta"  para mostrar información dentro de recuadros.
# Recibe un título, un contenido y un color para poder reutilizarla en diferentes partes del juego.
# El color tiene uno predeterminado, pero podemos cambiarlo cuando lo necesitemos.
def tarjeta(titulo, contenido, color="#8d3224"):  
   
    display(HTML(f'''<div style="max-width:740px;margin:12px auto;padding:20px 24px;border:1px solid #ead6c0;border-radius:18px;background:#fffaf3;color:#39281d;font-family:Arial,sans-serif;box-shadow:0 4px 14px #00000010">
    <div style="font-size:23px;font-weight:800;color:{color};margin-bottom:10px">{titulo}</div>
    <div style="font-size:16px;line-height:1.7">{contenido}</div></div>'''))

# Esta función reúne el panel del cliente y el menú de ingredientes.
# Al llamarla en cada intento evitamos repetir las mismas instrucciones de presentación.
def mostrar_estado(cliente, pedido, dinero, satisfechos, intento):  
  
    # Mostramos el pedido y los datos importantes de la partida.
    tarjeta(f"👩‍🍳 CLIENTE {cliente} DE 5", f"<b>Pedido:</b> 🍕 {pedido}<br><b>Intento:</b> {intento}/2 &nbsp;·&nbsp; ⏱️ <b>50 segundos</b><br>💰 Saldo: <b>{dinero} €</b> &nbsp;·&nbsp; ⭐ Clientes satisfechos: <b>{satisfechos}/5</b>")  

    # Recorremos el diccionario de ingredientes para mostrar cada número con su emoji y nombre.
    # join une todas las opciones en una sola línea, separándolas con una barra.
    opciones = " &nbsp; | &nbsp; ".join(f"<b>{n}</b> {iconos[ing]} {ing.title()}" for n, ing in ingredientes.items())  

    # Mostramos el menú y recordamos que el número 0 sirve para terminar la pizza.
    tarjeta("🧺 ELIGE TUS INGREDIENTES", opciones + "<br><i>Introduce los números de uno en uno. Pulsa 0 para hornear.</i>", "#985b15")  

# "dinero = 50" porque necesitamos una variable que guarde el saldo inicial del jugador.
# Luego sumamos y restamos según aciertos y fallos.
dinero = 50
# "clientes_satisfechos = 0" porque necesitamos contar cuántos clientes se han atendido bien.
# Empieza en 0 porque al principio nadie está satisfecho. Al final se compara con 5 para saber si se gana.
clientes_satisfechos = 0
# Convertimos las cuatro claves del diccionario de recetas en una lista de pedidos.
pedidos = list(recetas)  
# Mezclamos esa lista sin cambiar los ingredientes de las recetas.
random.shuffle(pedidos)  
# Añadimos una quinta pizza elegida al azar porque tenemos que atender a cinco clientes.
# Como solo hay cuatro recetas, una de ellas se repetirá.
pedidos.append(random.choice(list(recetas)))  

# Limpiamos la salida anterior para que la partida comience con una pantalla ordenada.
clear_output(wait=True)  

# Mostramos el título del juego con un diseño personalizado utilizando HTML.
display(HTML('''<div style="max-width:740px;margin:14px auto 20px;padding:30px 22px;border-radius:20px;background:linear-gradient(135deg,#842b24,#c65d31);text-align:center;color:white;font-family:Arial,sans-serif;box-shadow:0 7px 20px #00000025">
<div style="font-size:36px;font-weight:900">🍕 PIZZA PATRICCINI 🍕</div>
<div style="font-size:18px;margin:12px 0"><i>¡Ponte el delantal y enciende el horno!</i></div>
<div style="font-size:16px;font-weight:bold">⭐ NIVEL 1 · CINCO CLIENTES ⭐</div></div>'''))
tarjeta("📜 ¿CÓMO SE JUEGA?", "💰 Empiezas con <b>50 €</b>. Cada pizza correcta suma <b>20 €</b> y cada incorrecta resta <b>10 €</b>.<br>⏳ Tienes <b>50 segundos</b> por intento. Si se agota el tiempo, pierdes <b>5 €</b> y dispones de una segunda oportunidad.<br>🏆 Para ganar, debes acertar los <b>5 pedidos</b>.")  

# Utilizamos un FOR para recorrer los cinco pedidos y atender a cada cliente por orden.
# enumerate permite obtener tanto el número del cliente como la pizza que ha pedido.
# start=1 hace que los clientes se numeren del 1 al 5 en lugar de empezar por el 0.
for numero_cliente, pedido in enumerate(pedidos, start=1):  
   
    acertada = False
    
    intento = 1  
    # Usamos WHILE para permitir un máximo de dos intentos cuando se agota el tiempo.
    while intento <= 2:  
        # Mostramos al jugador su pedido, el saldo, los aciertos y los ingredientes disponibles.
        mostrar_estado(numero_cliente, pedido, dinero, clientes_satisfechos, intento)  
        # Lista vacía donde guardaremos los ingredientes elegidos en ESTE intento.
        pizza = []
        # Registramos el inicio del intento con un reloj apropiado para medir tiempo transcurrido.
        inicio = time.monotonic()  
        # Esta variable indica si se ha agotado el tiempo del intento.
        # Empieza en False y cambia a True cuando se cumplen los 50 segundos.
        agotado = False  
        # Usamos WHILE porque no sabemos cuántos ingredientes elegirá el jugador.
        # El bucle termina cuando pulsa 0 o se acaba el tiempo.
        while True:
            # Calculamos los segundos restantes y evitamos que aparezcan números negativos.
            restante = max(0, 50 - int(time.monotonic() - inicio))  
            # Si ya no queda tiempo, salimos del bucle de ingredientes.
            if restante == 0:  
                agotado = True
                break
            # input recoge la elección del teclado y strip elimina espacios al principio y al final.
            # strip() elimina espacios iniciales/finales (" 3 " pasa a "3").
            # f"...{restante}..." inserta el tiempo restante en el mensaje.
            entrada = input(f"⏳ {restante}s | Ingrediente (0 = hornear): ").strip()  
            # Volvemos a comprobar el tiempo después de recibir la respuesta.
            if time.monotonic() - inicio >= 50:  
                agotado = True
                break
            # isdigit comprueba si la entrada contiene números.
            # Así evitamos errores si el jugador escribe letras o deja la entrada vacía.
            if not entrada.isdigit():  
                print("⚠️ Escribe un número del 0 al 6.")  
                continue
            # Convertimos la entrada de texto a un número entero.
            # Lo necesitamos para buscar el ingrediente en el diccionario.
            opcion = int(entrada)  
            # El número 0 indica que ya ha terminado de preparar la pizza.
            # break sale del bucle de selección para comprobar la receta.
            if opcion == 0:
                break
            # Comprobamos que el número elegido exista en el menú de ingredientes.
            # Si no existe, mostramos un aviso y dejamos que el jugador vuelva a elegir.
            if opcion not in ingredientes:  
                print("⚠️ Ingrediente no disponible. Prueba otra vez.")  
                # continue vuelve al principio del WHILE DE INGREDIENTES:
                # no borra la pizza ni reinicia el temporizador de este intento.
                continue
            # Recuperamos el nombre del ingrediente correspondiente al número elegido.
            ingrediente = ingredientes[opcion]  
            # Añadimos el ingrediente a la lista de la pizza.
            # También mostramos su emoji y los ingredientes que se han seleccionado.
            pizza.append(ingrediente)
            print(f"✅ Añadido: {iconos[ingrediente]} {ingrediente.title()} | Tu pizza: {', '.join(pizza)}")  

        # Comprobamos si se ha agotado el tiempo antes de terminar la pizza.
        if agotado:
            dinero -= 5
            tarjeta("⌛ ¡SE ACABÓ EL TIEMPO!", f"Pierdes <b>5 €</b>. Saldo actual: <b>{dinero} €</b>.", "#b04c25")  
            # Solo el primer agotamiento permite una segunda oportunidad.
            if intento == 1:
                tarjeta("💛 SEGUNDA OPORTUNIDAD", "El cliente te concede un último intento. ¡Tú puedes!")  
                # Aumentamos el número de intento de 1 a 2.
                # += 1 equivale a intento = intento + 1.
                # La siguiente vuelta del while intento<=2 reinicia pizza e inicio.
                intento += 1  
                # continue vuelve al principio del WHILE DE INGREDIENTES:
                # no borra la pizza ni reinicia el temporizador de este intento.
                continue
            tarjeta("🚶 EL CLIENTE SE MARCHA", "No se ha podido entregar el pedido a tiempo.", "#b04c25")  
            break

        # Comparamos los ingredientes elegidos con los de la receta correcta.
        # sorted ordena ambas listas para compararlas sin importar el orden de selección.
        # También comprueba que no falten ingredientes ni haya ingredientes repetidos.
        if sorted(pizza) == sorted(recetas[pedido]):
            tarjeta("🔥 ¡PIZZA CORRECTA!", "<i>¡Qué buena pinta!</i> La pizza está en el horno...", "#23734e")  
            # Usamos FOR para hacer una cuenta atrás de tres segundos durante el horneado.
            # range(3, 0, -1) genera los números 3, 2 y 1.
            for segundos in range(3, 0, -1):  
                print(f"🔥 Horneando... {segundos}")  
                # Esperamos un segundo entre cada número para simular el horneado.
                time.sleep(1)
            # Si la pizza es correcta, sumamos 20 € al saldo.
            dinero += 20
            # Aumentamos en uno el número de clientes satisfechos.
            clientes_satisfechos += 1
            # Dejamos constancia de que el pedido se ha acertado.
            acertada = True
            tarjeta("🎉 ¡CLIENTE SATISFECHO!", f"Has ganado <b>20 €</b>. Saldo actual: <b>{dinero} €</b>.", "#23734e")  
        # else se ejecuta cuando la comparación anterior ha sido falsa.
        else:
            # Si la pizza es incorrecta, restamos 10 € al saldo.
            dinero -= 10
            tarjeta("❌ PIZZA INCORRECTA", f"Pierdes <b>10 €</b>. Saldo actual: <b>{dinero} €</b>.<br><i>Receta correcta:</i> {', '.join(recetas[pedido])}.", "#a63b34")  
        # Terminamos con este cliente después de comprobar la pizza.
        # Si la receta es incorrecta, no hay otro intento; solo se repite si se agota el tiempo.
        break

# Al terminar los cinco clientes, mostramos el resultado final de la partida.
# Comprobamos si se han conseguido los cinco pedidos para decidir si se gana o se pierde.
# La expresión condicional permite elegir el mensaje y el color según el resultado.
tarjeta("🏁 FIN DEL NIVEL 1", f"⭐ Clientes satisfechos: <b>{clientes_satisfechos}/5</b><br>💰 Dinero final: <b>{dinero} €</b><br>" + ("🏆 <b>¡NIVEL SUPERADO!</b> Eres una auténtica estrella de la pizza." if clientes_satisfechos == 5 else "💪 <b>NIVEL NO SUPERADO.</b> ¡Vuelve a intentarlo y consigue los cinco pedidos!"), "#23734e" if clientes_satisfechos == 5 else "#a63b34")  

#Aldara se encargó de desarrollar la base del juego mediante la ejecución de un cliente. 
#Posteriormente, Aiara amplió el proyecto a cinco clientes y modificó la variable de tiempo. 
#Patricia modificó la variable de dinero estableciendo el sistema de recompensas y penalizaciones.
#Finalmente, Eva fue la responsable del diseño estético y visual del juego.
```
